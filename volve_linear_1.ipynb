# Volve Production Data Analysis
This notebook replicates a RapidMiner workflow for predicting On-Stream Hours (`ON_STREAM_HRS`) using Linear Regression.

```
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge
from sklearn.metrics import root_mean_squared_error
```

## 1. Retrieve Data
In RapidMiner, this was the `Retrieve Volve_Production_Data4` block. 
Ensure you upload the dataset to your Colab environment or mount your Google Drive before running this cell.

```
# Replace the string below with the actual path to your CSV file in Google Colab.
file_path = "Volve_Production_Data4.csv" 
df = pd.read_csv(file_path)
```

## 2. Replace Missing Values
The RapidMiner process uses the `Replace Missing Values` operator set to the `average` default.
We will fill missing values with the mean of each respective column.

```
# Fill missing values with the average (mean) of the column
df = df.fillna(df.mean(numeric_only=True))
```

## 3. Set Role and Select Attributes
We set `ON_STREAM_HRS` as our target label. Then, we filter the dataset to only include the specific subset of attributes defined in the RapidMiner `Select Attributes` operator.

```
# Define the target variable (Label)
target = 'ON_STREAM_HRS'
y = df[target]

# Define the selected features
features = [
    'AVG_ANNULUS_PRESS', 
    'AVG_CHOKE_SIZE_P', 
    'AVG_DOWNHOLE_PRESSURE', 
    'AVG_DOWNHOLE_TEMPERATURE', 
    'AVG_DP_TUBING', 
    'AVG_WHP_P', 
    'AVG_WHT_P', 
    'BORE_GAS_VOL', 
    'BORE_OIL_VOL', 
    'BORE_WAT_VOL'
]
X = df[features]
```

## 4. Split Data
The RapidMiner `Split Data` operator partitions the data into 70% for training and 30% for testing.

```
# Split the data into training (70%) and testing (30%) sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1992)
```

## 5. Linear Regression Modeling
The RapidMiner process utilizes a `Linear Regression` operator. We noticed it included a `ridge` parameter set to `1.0E-8`. We will use Scikit-Learn's `Ridge` model to perfectly mirror this mathematical penalty.

```
# Initialize and train the model
model = Ridge(alpha=1e-8)
model.fit(X_train, y_train)
```

## 6. Apply Model & Performance
Finally, we apply the trained model to our unseen test data and evaluate its performance using Root Mean Squared Error (RMSE), which was the primary criterion checked in the RapidMiner `Performance` operator.

```
# Apply the model to predict the test set
y_pred = model.predict(X_test)

# Calculate Root Mean Squared Error
rmse = root_mean_squared_error(y_test, y_pred)
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
```
